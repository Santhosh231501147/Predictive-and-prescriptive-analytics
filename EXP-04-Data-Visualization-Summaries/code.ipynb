# Data Visualization, Histograms and Variable Summaries

**Objective:** Demonstrate data visualization, histograms, and summaries across multiple variables.

**Dataset:** Synthetic sales analytics dataset

This notebook is Colab-ready and saves tables, metrics, and visual outputs under
`results/`. Public datasets or compact sample datasets are used so the workflow
remains reproducible.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

RESULTS_DIR = Path("results")
RESULTS_DIR.mkdir(exist_ok=True)
rng = np.random.default_rng(42)
```

```python
df = pd.DataFrame(
    {
        "segment": rng.choice(["A", "B", "C", "D"], 600, p=[0.25, 0.35, 0.25, 0.15]),
        "spend": rng.gamma(2.2, 900, 600),
        "orders": rng.poisson(4, 600),
        "satisfaction": rng.normal(4.0, 0.55, 600),
    }
)
summary = df.groupby("segment").agg(
    avg_spend=("spend", "mean"),
    std_spend=("spend", "std"),
    avg_orders=("orders", "mean"),
    avg_satisfaction=("satisfaction", "mean"),
)
display(summary)
summary.to_csv(RESULTS_DIR / "variable_summary.csv")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.histplot(df["spend"], bins=35, kde=True, ax=axes[0])
axes[0].set_title("Spend Histogram")
summary["avg_spend"].plot(kind="bar", yerr=summary["std_spend"], ax=axes[1], color="teal")
axes[1].set_title("Average Spend by Segment")
axes[1].set_ylabel("Spend")
plt.tight_layout()
plt.savefig(RESULTS_DIR / "visualization_dashboard.png", dpi=180)
plt.show()
```
