# Descriptive Statistics and Distribution Analysis

**Objective:** Demonstrate mean, standard deviation, variance, normal and uniform distributions, and correlation.

**Dataset:** Synthetic business dataset

This notebook is Colab-ready and saves tables, metrics, and visual outputs under
`results/`. Public datasets or compact sample datasets are used so the workflow
remains reproducible.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

RESULTS_DIR = Path("results")
RESULTS_DIR.mkdir(exist_ok=True)
rng = np.random.default_rng(42)
```

```python
df = pd.DataFrame(
    {
        "sales": rng.normal(50000, 9000, 500),
        "profit": rng.normal(12000, 2500, 500),
        "discount": rng.uniform(0, 35, 500),
        "rating": rng.normal(4.1, 0.45, 500),
    }
)
df["profit"] = df["profit"] + 0.22 * (df["sales"] - df["sales"].mean())
stats = df.agg(["mean", "std", "var", "min", "max"]).T
corr = df.corr(numeric_only=True)
display(stats)
display(corr)
stats.to_csv(RESULTS_DIR / "descriptive_statistics.csv")
corr.to_csv(RESULTS_DIR / "correlation_matrix.csv")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].hist(df["sales"], bins=30, alpha=0.8, label="Sales")
axes[0].axvline(df["sales"].mean(), color="red", linestyle="--", label="Mean")
axes[0].set_title("Normal-like Sales Distribution")
axes[0].legend()
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, ax=axes[1])
axes[1].set_title("Correlation Matrix")
plt.tight_layout()
plt.savefig(RESULTS_DIR / "statistics_dashboard.png", dpi=180)
plt.show()
```
